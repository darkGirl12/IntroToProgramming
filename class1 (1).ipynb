```python
#Exercise3 - Even or Odd

#Get value from the user
num = int(input("Enter a number:"))

#Check if it is even or odd
if num%2==0:
    print ("The number ", num,"is even")
else:
    print ("The number ", num,"is odd")
```

```python
#Exercise4
# let the variable be an integer
integer = int(input("Enter an integer number:"))

#let it print div3 if the integer is divisible by 3
if integer%3==0:
    print("The integer ", integer,"div3")
else:
    print("The number ", integer,"is not divisible by 3")
```

```python
#For exercise2 using the for loops
numbers = int(input("How many numbers are you entering?"))

numberSum = 0
for i in range(numbers):
    numberSum+= int(input("Enter a number:"))

#Calculate the mean
mean = numberSum/numbers

#show result
print("The mean of the number is ", mean)
```

```python
#Exercise 8
num = int(input("Enter a value:"))

#Initialize the value
factorial = 1

#Calculate the factorial
for i in range(1, num+1):
    factorial *= i

print(f"Factorial of {num} is {factorial}")
```

```python
#Exercise 19
seconds = int(input('Enter the duration of seconds:'))

print(seconds)
```

```python
hours = seconds // 3600
remaining_seconds = seconds % 3600

minutes = remaining_seconds // 60
sec_seconds = remaining_seconds % 60

print(hours, 'Hours', minutes, 'Minutes', sec_seconds, 'Seconds')
```

```python
#Exercise 30
sales =  [120, 135, 98, 143, 160, 155, 170, 90, 130, 148, 165, 210]

total = 0
for value in sales:
    total+= value

average = total / len(sales)
print("Total: %d | Average: %.2f" %(total, average))

best = max(sales)
worst = min(sales)

print("The best month: %d (%d)" %(sales.index(best) + 1,best))
print("The worst month: %d (%d)" %(sales.index(worst) + 1,worst))

aboveAvg = []
for i in range (len(sales)):
    if sales[i] > average:
        aboveAvg.append(i+1)
print("The months above arange:", aboveAvg)

for i in range(1, len(sales)):
    growth = sales[i] - sales[i-1]/sales[i-1] * 100
    print("Month %d: %.1f" %(i + 1, growth))
```
